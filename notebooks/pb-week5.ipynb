```python
likes = [45, 120, 80, 200, 150]
max = likes[0]
second = likes[0]

for like in likes:
    if like > max:
        second = max
        max = like
    elif like > second and like != max:
        second = like

print(second)
```

```python
limit = 100
count = 0
for i in range(2, limit+1):
    for j in range(2, i):
        if i%j == 0: 
            break
    else:
        count += 1

print(count)
                
```

```python
limit = 100
count = 0
for i in range(2, limit+1):
    is_prime = True
    for j in range(2, i):
        if i%j == 0: 
            is_prime = False
            break
    if is_prime:
        count += 1

print(count)
                
```
